1. Write a function `is_palindrome(s: str) -> bool` that ignores case and spaces


```python
def is_palindrome(s:str) -> bool:
    return s.lower().strip().replace(" ","")[::-1] == s.lower().strip().replace(" ","")

is_palindrome('abcba')
```

2. Create `reverse_words(s: str) -> str` that reverses word order: "hello world" → "world hello"

```python
def reverse_words(s:str) -> str:
    return ' '.join(s.split()[::-1])

reverse_words('hello world')
```

3. Implement `title_case(s: str) -> str` without using `.title()`: "hello world" → "Hello World"


```python
def title_case(s:str) -> str:
    lst = []
    for i in s.split():
        lst.append(i[0].upper()+i[1::])
    return ' '.join(lst)

title_case('hello world')
        
```

1. `merge_sorted(list1: list[int], list2: list[int]) -> list[int]` - merge two sorted lists


```python
lst = [1,2,3,4,5,6,7,8,9]
def chunk_list(items: list[int], size: int) -> list[list[int]]:
    lst = []
    i = 0
    while i + size < len(items):
        lst.append(items[i:i+size])
        i += size
    lst.append(items[i::])

    return lst
 
print(chunk_list(lst,4))
```

```python
def flatten(nested: list[list[int]]) -> list[int]:
    lst: list[int] = []
    for i in nested:
        for j in i:
            lst.append(j)
    return lst

flatten([[1,2,4,4],[5,6]])
```

```python
def invert_dict(d: dict[str, str]) -> dict[str, str]:
    result: dict[str, str] = {}
    for key, value in d.items():
        result[value] = key
    return result

print(invert_dict({'a':'b','c':'d','e':'f'}))
```

```python

d1 = {'a':'b','c':'d','e':'f'}
d2 = {'z':'d','g':'v','w':'c'}

def merge_dicts(*dicts: dict[str,str]) -> dict[str, str]:
    result: dict[str, str] = {}
    for i in dicts:
        for j in i:
            result[j] = i[j]
    return result

merge_dicts(d1,d2)
```

```python
def common_elements(list1: list[int], list2: list[int]) -> set[int]:
    return set(list1).intersection(set(list2))

common_elements([1,2,3,4,5],[4,5,6,7])
```

```python
def build_query(**params) -> str:
    """Build URL query string from keyword arguments."""
    querry:str = '?'
    for key,value in params.items():
        querry += f'{key}={value}'
        querry += '&'
    if querry[-1] == '&':
        querry = querry[:len(querry)-2]
    return querry
 
print(build_query(page=1,limit=10))
```
